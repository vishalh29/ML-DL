```python
import nltk
# Download required NLTK data for lemmatization
nltk.download('wordnet')
nltk.download('omw-1.4')  # For additional language support

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "runner", "runs", "ran", "run","writer", "writing", "wrote", "writes","programming","programmer","program"]

print("Testing single word lemmatization:")
result = lemmatizer.lemmatize("running", pos="n")
print(f"running -> {result}")

```

```python
print("\nLemmatizing all words:")
for i in words:
    lemmatized = lemmatizer.lemmatize(i, pos="n")
    print(f"{i} -------> {lemmatized}")

```

```python
print("\nLemmatization with different POS tags:")
print("Note: pos='n' (noun), pos='v' (verb), pos='a' (adjective), pos='r' (adverb)")

# Demonstrate better lemmatization with correct POS tags
test_words = [
    ("running", "v"),  # verb form
    ("running", "n"),  # noun form
    ("better", "a"),   # adjective
    ("better", "r"),   # adverb
    ("writes", "v"),   # verb
    ("programming", "v"),  # verb
    ("programming", "n"),  # noun
]

for word, pos in test_words:
    lemmatized = lemmatizer.lemmatize(word, pos=pos)
    print(f"{word} (pos='{pos}') -------> {lemmatized}")
```
