```python
#stemming

words = ["history","running", "runner", "runs", "ran", "run","writer", "writing", "wrote", "writes","programming","programmer","program"]
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

```

```python
for i in words:
    print(i+ "------>"+stemmer.stem(i))
```

```python
stemmer.stem("congratulations")
```

```python
from nltk.stem import RegexpStemmer
```

```python
regexp_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
print(regexp_stemmer.stem("running"))
print(regexp_stemmer.stem("writer"))
print(regexp_stemmer.stem("writing"))
print(regexp_stemmer.stem("programmer"))
print(regexp_stemmer.stem("program"))
print(regexp_stemmer.stem("congratulations"))
print(regexp_stemmer.stem("eating"))
```

```python
from nltk.stem import SnowballStemmer

snowball_stemmer = SnowballStemmer('english')

```

```python
for i in words:
    print(i+ "------>"+snowball_stemmer.stem(i))
```
